```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, to_timestamp
import pandas as pd

spark = SparkSession.builder \
    .appName("EcoData-Extraccion") \
    .master("local[*]") \
    .getOrCreate()

# ============================================
# 1. Leer con pandas saltando las 2 primeras filas
# ============================================
pdf = pd.read_csv(
    "/opt/ecodata/data/juliaca_clima_historico.csv",
    skiprows=2                # Saltamos metadatos
)

print("Columnas originales:")
print(pdf.columns.tolist())
print(f"\nRegistros: {len(pdf)}")
print(pdf.head(3))
```

```python
# Limpiar nombres de columnas
pdf.columns = [c.replace(" (°C)", "")
                .replace(" (km/h)", "")
                .replace(" (°)", "")
                .replace(" (%)", "")
                .replace(" ", "_")
                .lower() for c in pdf.columns]

# Convertir a Spark
df_clima = spark.createDataFrame(pdf)

# Convertir time a timestamp
df_clima = df_clima.withColumn("time", to_timestamp(col("time")))

# Seleccionar columnas útiles
df_clima = df_clima.select(
    "time",
    "temperature_2m",
    "wind_speed_10m",
    "wind_direction_10m",
    "relative_humidity_2m"
)

print("\n=== CLIMA LIMPIO EN SPARK ===")
df_clima.printSchema()
df_clima.show(10, truncate=False)
print(f"Total registros: {df_clima.count()}")
```

```python
# ============================================
# Cargar y limpiar CALIDAD DEL AIRE
# ============================================
pdf_aire = pd.read_csv(
    "/opt/ecodata/data/juliaca_calidad_aire.csv",
    skiprows=2
)

print("Columnas originales de calidad del aire:")
print(pdf_aire.columns.tolist())
print(f"Registros: {len(pdf_aire)}")
print(pdf_aire.head(3))
```

```python
from pyspark.sql.functions import col, to_timestamp

# ============================================
# Limpiar nombres de columnas
# ============================================
pdf_aire.columns = [
    c.replace(" (μg/m³)", "")
     .replace(" (EAQI)", "")
     .replace(" (USAQI)", "")
     .replace(" ", "_")
     .lower()
    for c in pdf_aire.columns
]

print("Columnas limpias:", pdf_aire.columns.tolist())

# Convertir a Spark
df_aire = spark.createDataFrame(pdf_aire)

# Convertir tipos
df_aire = (df_aire
    .withColumn("time", to_timestamp(col("time")))
    .withColumn("pm10", col("pm10").cast("double"))
    .withColumn("pm2_5", col("pm2_5").cast("double"))
    .withColumn("european_aqi", col("european_aqi").cast("double"))
    .withColumn("us_aqi", col("us_aqi").cast("double"))
)

# Seleccionar solo columnas útiles
df_aire = df_aire.select(
    "time",
    "pm10",
    "pm2_5",
    "european_aqi",
    "us_aqi"
)

print("\n=== CALIDAD DEL AIRE LIMPIA ===")
df_aire.printSchema()
df_aire.show(5, truncate=False)
print(f"Total registros: {df_aire.count()}")
```

```python
from pyspark.sql.functions import col, count, when, isnan, trim, year, month

# ============================================
# 1. NULOS – Clima
# ============================================
print("=== NULOS EN CLIMA ===")
df_clima.select([
    count(when(col(c).isNull(), c)).alias(c) 
    for c in df_clima.columns
]).show()

# ============================================
# 2. NULOS – Calidad del aire
# ============================================
print("=== NULOS EN CALIDAD DEL AIRE ===")
df_aire.select([
    count(when(col(c).isNull(), c)).alias(c) 
    for c in df_aire.columns
]).show()

# ============================================
# 3. DUPLICADOS – por columna time
# ============================================
print("=== DUPLICADOS CLIMA (por time) ===")
dup_clima = df_clima.groupBy("time").count().filter("count > 1")
print(f"Timestamps duplicados en clima: {dup_clima.count()}")

print("\n=== DUPLICADOS AIRE (por time) ===")
dup_aire = df_aire.groupBy("time").count().filter("count > 1")
print(f"Timestamps duplicados en aire: {dup_aire.count()}")

# ============================================
# 4. ELIMINAR DUPLICADOS (por si acaso)
# ============================================
df_clima = df_clima.dropDuplicates(["time"])
df_aire  = df_aire.dropDuplicates(["time"])

print(f"\nClima después de dropDuplicates: {df_clima.count()}")
print(f"Aire  después de dropDuplicates: {df_aire.count()}")

# ============================================
# 5. FILTRADO de rango físico (outliers)
# ============================================
df_clima = df_clima.filter(
    col("temperature_2m").between(-20, 40) &
    col("wind_speed_10m").between(0, 100) &
    col("relative_humidity_2m").between(0, 100)
)

df_aire = df_aire.filter(
    col("pm10").between(0, 500) &
    col("pm2_5").between(0, 300)
)

print(f"\nClima después de filtrado físico: {df_clima.count()}")
print(f"Aire  después de filtrado físico: {df_aire.count()}")
```

```python
from pyspark.sql.functions import year, month, col

# ============================================
# 1. Agregar columnas de partición
# ============================================
df_clima_gold = (df_clima
    .withColumn("año", year(col("time")))
    .withColumn("mes", month(col("time")))
)

df_aire_gold = (df_aire
    .withColumn("año", year(col("time")))
    .withColumn("mes", month(col("time")))
)

# ============================================
# 2. Escribir CLIMA particionado
# ============================================
ruta_clima = "/opt/ecodata/artifacts/juliaca_clima_particionado"

df_clima_gold.write \
    .mode("overwrite") \
    .partitionBy("año", "mes") \
    .parquet(ruta_clima)

print(f"✅ Clima escrito en: {ruta_clima}")

# ============================================
# 3. Escribir CALIDAD DEL AIRE particionada
# ============================================
ruta_aire = "/opt/ecodata/artifacts/juliaca_aire_particionado"

df_aire_gold.write \
    .mode("overwrite") \
    .partitionBy("año", "mes") \
    .parquet(ruta_aire)

print(f"✅ Aire escrito en: {ruta_aire}")

# ============================================
# 4. Verificar lectura de vuelta (integridad)
# ============================================
df_clima_leido = spark.read.parquet(ruta_clima)
df_aire_leido  = spark.read.parquet(ruta_aire)

print(f"\nClima leído de vuelta: {df_clima_leido.count()} (esperado: {df_clima.count()})")
print(f"Aire  leído de vuelta: {df_aire_leido.count()}  (esperado: {df_aire.count()})")

# ============================================
# 5. Verificar PartitionFilters (optimización)
# ============================================
print("\n=== Plan de ejecución con filtro de partición (clima 2024-03) ===")
df_clima_leido.filter((col("año") == 2024) & (col("mes") == 3)).explain(True)
```

```python
from pyspark.sql.functions import col, year, month, hour, dayofweek
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.regression import LinearRegression, RandomForestRegressor
from pyspark.ml.evaluation import RegressionEvaluator
from pyspark.ml import Pipeline

# ============================================
# Cargar la capa Gold de clima
# ============================================
df = spark.read.parquet("/opt/ecodata/artifacts/juliaca_clima_particionado")

print(f"Registros cargados: {df.count()}")
df.printSchema()
df.show(5)
```

```python
from pyspark.sql.functions import col, hour, month

# ============================================
# 1. Crear features adicionales de tiempo
# ============================================
df_ml = (df
    .withColumn("hour", hour(col("time")))
    .withColumn("mes", month(col("time")))   # ya existe, pero lo reforzamos
)

# ============================================
# 2. Definir predictores y objetivo
# ============================================
PREDICTORES = [
    "wind_speed_10m",
    "wind_direction_10m",
    "relative_humidity_2m",
    "hour",
    "mes"
]

OBJETIVO = "temperature_2m"

# Nos quedamos solo con las columnas necesarias y sin nulos
df_ml = df_ml.select(PREDICTORES + [OBJETIVO]).na.drop()

print(f"Registros listos para ML: {df_ml.count()}")
df_ml.show(5)
```

```python
from pyspark.ml.feature import VectorAssembler

# ============================================
# 3. Ensamblar vector de features
# ============================================
assembler = VectorAssembler(
    inputCols=PREDICTORES,
    outputCol="features"
)

df_vector = assembler.transform(df_ml).select("features", col(OBJETIVO).alias("label"))

print("=== Datos con vector de features ===")
df_vector.show(5, truncate=False)

# ============================================
# 4. Split entrenamiento / prueba (80/20)
# ============================================
df_train, df_test = df_vector.randomSplit([0.8, 0.2], seed=42)

print(f"Train: {df_train.count()}  |  Test: {df_test.count()}")
```

```python
from pyspark.ml.regression import LinearRegression
from pyspark.ml.evaluation import RegressionEvaluator

# ============================================
# 5. Modelo base (sin regularización fuerte)
# ============================================
lr_base = LinearRegression(
    featuresCol="features",
    labelCol="label",
    maxIter=20,
    regParam=0.0,          # sin regularización
    elasticNetParam=0.0
)

modelo_base = lr_base.fit(df_train)

# Predicciones sobre test
predicciones_base = modelo_base.transform(df_test)

print("=== Primeras predicciones ===")
predicciones_base.select("label", "prediction").show(8)

# ============================================
# 6. Métricas de evaluación
# ============================================
evaluator_rmse = RegressionEvaluator(labelCol="label", predictionCol="prediction", metricName="rmse")
evaluator_r2   = RegressionEvaluator(labelCol="label", predictionCol="prediction", metricName="r2")
evaluator_mae  = RegressionEvaluator(labelCol="label", predictionCol="prediction", metricName="mae")

rmse_base = evaluator_rmse.evaluate(predicciones_base)
r2_base   = evaluator_r2.evaluate(predicciones_base)
mae_base  = evaluator_mae.evaluate(predicciones_base)

print("\n=== MÉTRICAS MODELO BASE ===")
print(f"RMSE : {rmse_base:.4f}")
print(f"R²   : {r2_base:.4f}")
print(f"MAE  : {mae_base:.4f}")

# Coeficientes
print("\n=== Coeficientes del modelo ===")
print(f"Intercepto: {modelo_base.intercept:.4f}")
for nombre, coef in zip(PREDICTORES, modelo_base.coefficients):
    print(f"  {nombre}: {coef:.6f}")
```

```python
from pyspark.ml.regression import RandomForestRegressor

# ============================================
# 7. Tres configuraciones de LinearRegression
# ============================================
configs = [
    {"nombre": "LR_sin_reg",     "regParam": 0.0,  "elasticNetParam": 0.0},
    {"nombre": "LR_Ridge",       "regParam": 0.1,  "elasticNetParam": 0.0},   # L2
    {"nombre": "LR_ElasticNet",  "regParam": 0.1,  "elasticNetParam": 0.5},   # mezcla L1+L2
]

resultados = []

for cfg in configs:
    lr = LinearRegression(
        featuresCol="features",
        labelCol="label",
        maxIter=20,
        regParam=cfg["regParam"],
        elasticNetParam=cfg["elasticNetParam"]
    )
    modelo = lr.fit(df_train)
    pred = modelo.transform(df_test)
    
    rmse = evaluator_rmse.evaluate(pred)
    r2   = evaluator_r2.evaluate(pred)
    mae  = evaluator_mae.evaluate(pred)
    
    resultados.append({
        "modelo": cfg["nombre"],
        "rmse": round(rmse, 4),
        "r2": round(r2, 4),
        "mae": round(mae, 4)
    })
    print(f"{cfg['nombre']:20s} → RMSE={rmse:.4f}  R²={r2:.4f}  MAE={mae:.4f}")

# ============================================
# 8. RandomForestRegressor
# ============================================
rf = RandomForestRegressor(
    featuresCol="features",
    labelCol="label",
    numTrees=30,
    maxDepth=8,
    seed=42
)

modelo_rf = rf.fit(df_train)
pred_rf = modelo_rf.transform(df_test)

rmse_rf = evaluator_rmse.evaluate(pred_rf)
r2_rf   = evaluator_r2.evaluate(pred_rf)
mae_rf  = evaluator_mae.evaluate(pred_rf)

resultados.append({
    "modelo": "RandomForest",
    "rmse": round(rmse_rf, 4),
    "r2": round(r2_rf, 4),
    "mae": round(mae_rf, 4)
})

print(f"{'RandomForest':20s} → RMSE={rmse_rf:.4f}  R²={r2_rf:.4f}  MAE={mae_rf:.4f}")

# ============================================
# 9. Tabla comparativa
# ============================================
print("\n=== TABLA COMPARATIVA DE MODELOS ===")
df_resultados = spark.createDataFrame(resultados)
df_resultados.orderBy("rmse").show(truncate=False)

# Importancia de variables del RandomForest
print("\n=== Importancia de variables (RandomForest) ===")
for nombre, imp in zip(PREDICTORES, modelo_rf.featureImportances):
    print(f"  {nombre:25s}: {imp:.4f}")
```

```python
# ============================================
# 10. Guardar el Modelo Seleccionado (Ganador: Random Forest)
# ============================================
ruta_modelo = "/opt/ecodata/models/clima_rf_model"

# Guardamos el modelo sobrescribiendo si la ruta ya existe
modelo_rf.write().overwrite().save(ruta_modelo)

print(f"\n✅ Sesión 4 Completada. Modelo RF guardado exitosamente en: {ruta_modelo}")
```
